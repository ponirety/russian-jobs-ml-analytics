```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/raw/hh_vacancies_2026_07.csv', encoding='utf-8')

df['target'] = df[['salary_from_rub', 'salary_to_rub']].mean(axis=1)
df['target'] = df['target'].fillna(df['salary_from_rub'])
df['target'] = df['target'].fillna(df['salary_to_rub'])

#df.shape[0] 129464

df = df[df['target'].notna()]
df = df.reset_index(drop=True)

df['target'].head(15)
#df.shape[0] 103146

columns_to_drop = [
    'salary_from', 'salary_to', 'currency', 'gross', 'salary_mode', 
    'salary_from_rub', 'salary_to_rub', 'salary_period', 'salary_shown',
    'vacancy_id', 'collected_area_id', 'role_ids', 'address_display',
    'responses_count', 'total_responses_count', 'publication_time', 'snapshot_date',
    'company', 'address_city', 'work_formats', 'work_schedule', 'working_hours',
    'languages', 'internship', 'night_shifts', 'response_letter_required', 'civil_law_contracts'
]
df = df.drop(columns=columns_to_drop)

df.to_csv('../data/processed/cleaned_vacancies.csv', index=False)
```
