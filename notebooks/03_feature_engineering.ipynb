```python
import pandas as pd
import numpy as np

df =  pd.read_csv('../data/processed/cleaned_vacancies.csv', encoding='utf-8')

columns = [
    'accept_temporary','accept_labor_contract', 'is_on_site', 'is_remote', 'is_hybrid', 'is_field_work'
]
df[columns] = df[columns].astype(int)

exp_mapping = {'Без опыта':0, '1-3 года':1, '3-6 лет':2, 'Более 6 лет':3}

df['experience'] = df['experience'].map(exp_mapping)

df['name_length'] = df['name'].str.len()
df['is_senior'] = df['name'].str.contains('senior|lead', case=False, na=False).astype(int)
df['is_junior'] = df['name'].str.contains('junior|intern', case=False, na=False).astype(int)

df = df.drop(columns='name')

df = pd.get_dummies(df, columns=['area_name', 'employment_form', 'category'], drop_first=True)
display(df.head())

df.to_csv('../data/processed/ready_for_ml.csv', index=False)
```
